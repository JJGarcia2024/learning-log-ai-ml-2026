```python
# Query to count the number of transactions per browser
query = """
        SELECT device.browser AS device_browser,
            SUM(totals.transactions) as total_transactions
        FROM `bigquery-public-data.google_analytics_sample.ga_sessions_20170801`
        GROUP BY device_browser
        ORDER BY total_transactions DESC
        """

# Run the query, and return a pandas DataFrame
# create_bqstorage_client=False avoids needing bigquery.readsessions.create permission
result = client.query(query).result().to_dataframe(create_bqstorage_client=False)
result.head()
```

```python
# Query to determine most popular landing point on the website

query = """
        -- hits.page.pagePath is a deeply nested field: hits (REPEATED RECORD) 
        -- -> page (RECORD) -> pagePath (STRING). Three levels deep.
        SELECT hits.page.pagePath as path,
            COUNT(hits.page.pagePath) as counts

        -- hits is a REPEATED RECORD (an array). You can't filter or select from it directly.
        -- UNNEST(hits) explodes each element of the hits array into its own separate row,
        -- aliased back as "hits" so dot notation like hits.type and hits.page.pagePath works.
        FROM `bigquery-public-data.google_analytics_sample.ga_sessions_20170801`, 
            UNNEST(hits) as hits

        -- hits.type = "PAGE" keeps only page view events (not clicks, events, or transactions).
        -- hits.hitNumber = 1 keeps only the FIRST hit of each session — the landing page.
        -- Together these two filters isolate exactly where each visitor entered the site.
        WHERE hits.type="PAGE" and hits.hitNumber=1

        -- Group rows by page path so COUNT() adds up sessions per landing page.
        GROUP BY path

        -- Most-visited landing pages appear first.
        ORDER BY counts DESC
        """

# Run the query and return a pandas DataFrame.
# create_bqstorage_client=False uses the standard REST API instead of the
# BigQuery Storage API, which requires an extra permission we don't have.
result = client.query(query).result().to_dataframe(create_bqstorage_client=False)
result.head()
```
